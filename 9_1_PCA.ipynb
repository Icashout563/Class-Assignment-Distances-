```python
#PCA on Iris
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

iris = pd.read_csv('iris.csv')
X_iris = iris.drop(columns=['variety'])
y_iris = iris['variety']

X_iris_scaled = StandardScaler().fit_transform(X_iris)

pca_iris = PCA(n_components=2)
iris_pcs = pca_iris.fit_transform(X_iris_scaled)

print("Explained variance ratio:", pca_iris.explained_variance_ratio_)

plt.figure(figsize=(8, 6))
for sp in y_iris.unique():
    mask = y_iris == sp
    plt.scatter(iris_pcs[mask, 0], iris_pcs[mask, 1], label=sp, edgecolor='k', alpha=0.8)
plt.xlabel(f'PC1 ({pca_iris.explained_variance_ratio_[0]*100:.1f}% variance)')
plt.ylabel(f'PC2 ({pca_iris.explained_variance_ratio_[1]*100:.1f}% variance)')
plt.title('PCA of Iris Dataset')
plt.legend(title='Species')
plt.show()
```

```python
#PCA on Diabetes
diabetes = pd.read_csv('diabetes.csv')
X_diab = diabetes.drop(columns=['Outcome'])
y_diab = diabetes['Outcome']

X_diab_scaled = StandardScaler().fit_transform(X_diab)

pca_diab = PCA(n_components=2)
diab_pcs = pca_diab.fit_transform(X_diab_scaled)

print("Explained variance ratio:", pca_diab.explained_variance_ratio_)

plt.figure(figsize=(8, 6))
for oc, lab in zip([0, 1], ['No Diabetes (0)', 'Diabetes (1)']):
    mask = y_diab == oc
    plt.scatter(diab_pcs[mask, 0], diab_pcs[mask, 1], label=lab, edgecolor='k', alpha=0.7)
plt.xlabel(f'PC1 ({pca_diab.explained_variance_ratio_[0]*100:.1f}% variance)')
plt.ylabel(f'PC2 ({pca_diab.explained_variance_ratio_[1]*100:.1f}% variance)')
plt.title('PCA of Diabetes Dataset')
plt.legend(title='Outcome')
plt.show()
```

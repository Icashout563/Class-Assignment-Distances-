```python
import pandas as pd
import numpy as np

df = pd.read_csv('distance_dataset.csv', index_col=0)
X = df['X'].values
Y = df['Y'].values
Z = df['Z'].values

#Task 1
manhattan_xy = np.abs(X - 6) + np.abs(Y - 6)
print(manhattan_xy)

```

```python
#Task 2
euclidean_xy = np.sqrt((X - 6)**2 + (Y - 6)**2)
print(euclidean_xy)
```

```python
#Task 3
p = 1.5
minkowski_xy = (np.abs(X - 6)**p + np.abs(Y - 6)**p) ** (1/p)
print(minkowski_xy)
```

```python
#Task 4
p = 1.5
minkowski_xyz = (np.abs(X - 6)**p + np.abs(Y - 6)**p + np.abs(Z - 6)**p) ** (1/p)
print(minkowski_xyz)
```

```python
#Task 5
import matplotlib.pyplot as plt

fig, axes = plt.subplots(2, 2, figsize=(12, 9))

axes[0,0].hist(manhattan_xy, bins=40)
axes[0,0].set_title('Manhattan Distance (X,Y) from (6,6)')

axes[0,1].hist(euclidean_xy, bins=40)
axes[0,1].set_title('Euclidean Distance (X,Y) from (6,6)')

axes[1,0].hist(minkowski_xy, bins=40)
axes[1,0].set_title('Minkowski (p=1.5) Distance (X,Y) from (6,6)')

axes[1,1].hist(minkowski_xyz, bins=40)
axes[1,1].set_title('Minkowski (p=1.5) Distance (X,Y,Z) from (6,6,6)')

plt.tight_layout()
plt.show()
```
